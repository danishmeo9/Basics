# **Creating Pyramids**

```python
for i in range(1,6):
    x="*"
    x=x * i
    print(f'{x:^10}')
```

```python
n = 5

# 1. Simple (left-aligned) Pyramid
print("1. Simple Pyramid")
for i in range(1, n+1):
    print('*' * i)

# 2. Inverted Pyramid
print("\n2. Inverted Pyramid")
for i in range(n, 0, -1):
    print('*' * i)

# 3. Centered (Equilateral) Pyramid
print("\n3. Centered Pyramid")
for i in range(1, n+1):
    print(' ' * (n-i) + '*' * (2*i-1))


    print()
```
